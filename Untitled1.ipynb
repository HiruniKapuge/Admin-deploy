```python
import cv2
from google.colab.patches import cv2_imshow
```

```python
#Read the image from file
image = cv2.imread('image_1.jpg')
```

```python
#Display the image in a window
cv2_imshow(image)
```

```python
#load the image
image = cv2.imread('image_1.jpg')
#Resize the image to 600*400
resized_image = cv2.resize(image,(600,400))
#save the resized image
cv2.imwrite('image.jpeg', resized_image)

cv2_imshow(resized_image)
```

```python
#load the image
image = cv2.imread('image_1.jpg')
 #get the images's height and width
(h,w) = image.shape[:2]
#Define the center of the image(using "//" round the result down to the nears whole number)
center = (w // 2, h//2)
#creates a rotation matrix that rotates the image 90 degrees counterclockwise around its center.
#1.0 - keep original size
matrix = cv2.getRotationMatrix2D(center,180,1.0)
```

```python
#oerform the rotation
rotated_image = cv2.warpAffine(image, matrix, (w,h))
#save the rotated image
cv2.imwrite('rotated_image.jpeg',rotated_image)


cv2_imshow(rotated_image)
```

```python
#load the image
image = cv2.imread('image_1.jpg')
#Apply Gaussian Blur
blurred_image = cv2.GaussianBlur(image,(15,15),0)
#Save the blurred image
cv2.imwrite('blurred_image.jpg',blurred_image)

cv2_imshow(blurred_image)
```

```python
#load the image
image = cv2.imread('image_1.jpg')
#Apply Gaussian Blur
blurred_image = cv2.GaussianBlur(image,(21,21),0)
#Save the blurred image
cv2.imwrite('blurred_image.jpg',blurred_image)

cv2_imshow(blurred_image)
```

```python
#Read image from file
image_1 = cv2.imread('image_1.jpg')
#convert to gray
gray_image = cv2.cvtColor(image_1, cv2.COLOR_BGR2GRAY)
#apply canny edge detector
edges = cv2.Canny(gray_image, 100, 200)
#Save the gray scale image
cv2.imwrite('edges.jpg', gray_image)
```
